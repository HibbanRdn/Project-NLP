## 📌 LANGKAH 3 — Buat Struktur Folder Proyek

```python
# Daftar semua folder yang dibutuhkan
FOLDERS = [
    "data/raw",
    "data/processed",
    "notebooks",
    "src",
    "models",
    "results",
    "app",
    "docs",
]

print("🔧 Membuat struktur folder...\n")
for folder in FOLDERS:
    path = os.path.join(ROOT, folder)
    os.makedirs(path, exist_ok=True)
    print(f"  ✅ {path}")

print("\n🎉 Semua folder berhasil dibuat!")
```

```python
# Verifikasi struktur folder
print(f"\n📁 Struktur folder di {ROOT}:\n")
for dirpath, dirnames, filenames in os.walk(ROOT):
    level = dirpath.replace(ROOT, '').count(os.sep)
    indent = '  ' * level
    print(f"{indent}📂 {os.path.basename(dirpath)}/")
    subindent = '  ' * (level + 1)
    for f in filenames:
        print(f"{subindent}📄 {f}")
```

## 📌 LANGKAH 5 — Verifikasi Instalasi

```python
print("🔍 Mengecek semua library...\n")

checks = {}

# Cek pandas
try:
    import pandas as pd
    checks["pandas"] = pd.__version__
except ImportError:
    checks["pandas"] = "❌ GAGAL"

# Cek numpy
try:
    import numpy as np
    checks["numpy"] = np.__version__
except ImportError:
    checks["numpy"] = "❌ GAGAL"

# Cek scikit-learn
try:
    import sklearn
    checks["scikit-learn"] = sklearn.__version__
except ImportError:
    checks["scikit-learn"] = "❌ GAGAL"

# Cek PySastrawi
try:
    from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
    checks["PySastrawi"] = "✅ OK"
except ImportError:
    checks["PySastrawi"] = "❌ GAGAL"

# Cek transformers
try:
    import transformers
    checks["transformers"] = transformers.__version__
except ImportError:
    checks["transformers"] = "❌ GAGAL"

# Cek sentence-transformers
try:
    import sentence_transformers
    checks["sentence-transformers"] = sentence_transformers.__version__
except ImportError:
    checks["sentence-transformers"] = "❌ GAGAL"

# Cek torch + GPU
try:
    import torch
    gpu_info = "✅ GPU tersedia" if torch.cuda.is_available() else "⚠️ CPU only (aktifkan GPU di Runtime)"
    checks["torch"] = f"{torch.__version__} — {gpu_info}"
except ImportError:
    checks["torch"] = "❌ GAGAL"

# Tampilkan hasil
print(f"{'Library':<25} {'Status / Versi'}")
print("-" * 55)
for lib, status in checks.items():
    icon = "✅" if "GAGAL" not in status else "❌"
    print(f"  {lib:<23} {status}")

print("\n🎉 Verifikasi selesai!")
```

## 📌 LANGKAH 7 — Test Cepat PySastrawi

Pastikan stemmer Bahasa Indonesia berjalan dengan benar.

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

# Init stemmer
stemmer_factory = StemmerFactory()
stemmer         = stemmer_factory.create_stemmer()

# Init stopword remover
sw_factory      = StopWordRemoverFactory()
stopword_list   = sw_factory.get_stop_words()

# Test
test_kalimat = "Implementasi pembelajaran mesin untuk mengklasifikasikan data teks berbahasa Indonesia"
hasil_stem   = stemmer.stem(test_kalimat)

print("🔤 Input   :", test_kalimat)
print("✂️  Stemmed :", hasil_stem)
print(f"\n📋 Total stopword Sastrawi: {len(stopword_list)} kata")
print("✅ PySastrawi berjalan dengan baik!")
```
